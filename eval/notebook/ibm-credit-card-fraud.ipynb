# IBM Credit Card Dataset Feasibility Assessment for Graph Machine Learning

## Dataset Overview

The IBM TabFormer credit card dataset is a synthetic dataset designed for fraud detection research. It contains realistic credit card transaction data with the following key characteristics:

### Dataset Structure
- **Source**: [IBM/TabFormer GitHub Repository](https://github.com/IBM/TabFormer/tree/main/data/credit_card)
- **Type**: Synthetic credit card transaction data
- **Purpose**: Fraud detection and financial analysis research
- **Format**: Tabular data suitable for graph transformation

### Key Features
- **Transaction ID**: Unique identifier for each transaction
- **Customer ID**: Unique identifier for each customer
- **Merchant ID**: Unique identifier for each merchant  
- **Transaction Amount**: Monetary value of the transaction
- **Timestamp**: Date and time of transaction
- **Fraud Label**: Binary indicator (1 = fraudulent, 0 = legitimate)

## Graph Representation Potential

### Proposed Graph Structure

**Bipartite Graph Design:**
- **Nodes**: 
  - Customer nodes (identified by Customer ID)
  - Merchant nodes (identified by Merchant ID)
- **Edges**: 
  - Transaction edges connecting customers to merchants
- **Edge Attributes**:
  - Transaction amount (weight)
  - Timestamp (temporal information)
  - Fraud label (target variable)

### Graph Machine Learning Opportunities

1. **Fraud Detection**
   - Anomaly detection in transaction patterns
   - Graph neural networks for fraud classification
   - Temporal graph analysis for suspicious behavior

2. **Network Analysis**
   - Community detection to identify merchant clusters
   - Customer-merchant relationship analysis
   - Transaction flow patterns

3. **Link Prediction**
   - Predicting future customer-merchant interactions
   - Risk assessment for new transaction patterns

## Feasibility Assessment

### ✅ Advantages
- **Rich relational structure**: Customer-merchant interactions naturally form a graph
- **Temporal dynamics**: Time-series nature enables temporal graph analysis
- **Real-world relevance**: Practical application in fraud detection
- **Labeled data**: Fraud labels enable supervised learning approaches
- **Synthetic nature**: Privacy-safe for experimentation

### ⚠️ Considerations
- **Synthetic data**: May not capture real-world complexity
- **Dataset size**: Need to verify scale for meaningful graph analysis
- **Feature engineering**: May require additional feature extraction
- **Computational requirements**: Graph ML algorithms can be resource-intensive

```python
# Get basic statistics about the dataset
stats = loader.get_basic_stats()

print("Dataset Statistics:")
print("=" * 20)
for key, value in stats.items():
    if isinstance(value, dict):
        print(f"\n{key}:")
        for sub_key, sub_value in value.items():
            print(f"  {sub_key}: {sub_value}")
    else:
        print(f"{key}: {value}")

# Show fraud distribution
print(f"\nFraud Rate: {stats['fraud_rate']:.2%}")
print(f"Normal Transactions: {stats['total_transactions'] - int(stats['total_transactions'] * stats['fraud_rate']):,}")
print(f"Fraud Transactions: {int(stats['total_transactions'] * stats['fraud_rate']):,}")

# Percentage breakdown appended after summary
total_tx = stats['total_transactions']
fraud_pct = stats['fraud_rate'] * 100
normal_pct = 100 - fraud_pct
print("\nPercentages:")
print(f"Normal: {normal_pct:.2f}%")
print(f"Fraud: {fraud_pct:.2f}%")
```
